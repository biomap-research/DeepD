# S288C chromosome III protein-synteny demo

This notebook reproduces the formal `AIYeast00` versus S288C chromosome III plot from the supplied DAGchainer anchors. Runtime code is consolidated in `synteny_demo.py`.


## 1. Locate or download runtime assets

Local assets are used when present. If they are absent, `ensure_demo_assets` downloads the `data/` subtree from the Hugging Face dataset `biomap-research/DeepD` (`synteny/` prefix). Override with `DEEPD_DATA_REPO_ID` or function arguments only if you host a fork.


```python
from pathlib import Path

from synteny_demo import ensure_demo_assets, audit_synteny_inputs, run_synteny_demo

cwd = Path.cwd().resolve()
ROOT = next(
    candidate for candidate in (cwd, cwd / 'synteny', cwd.parent)
    if (candidate / 'synteny_demo.py').is_file()
)
asset_status = ensure_demo_assets(ROOT)
print(f'Demo root: {ROOT}')
print(f"Assets downloaded in this run: {asset_status['downloaded']}")
```

## 2. Audit the formal upstream result

The notebook reads the recorded BLASTP and DAGchainer parameters and counts every supplied alignment and anchor row. It does not recompute or filter anchors.


```python
audit = audit_synteny_inputs(ROOT)
for key, value in audit.items():
    print(f'{key}: {value}')
assert audit['anchor_count'] == 163
```

## 3. Generate the formal plot

The consolidated module uses the original full-chromosome plotting implementation and preserves the production figure parameters, structural markers, and all 163 anchors.


```python
report = run_synteny_demo(ROOT, download_if_missing=False)
print(f"Status: {report['status']}")
print(f"Samples: {report['samples']}")
print(f"Anchors: {report['anchor_count']}")
for pdf in report['pdfs']:
    print(f'PDF: {pdf}')
```

## 4. Outputs

Generated files are written under `results/formal/`. The PDF is the canonical figure, and the adjacent `.summary.tsv` records the plotted coordinates, feature counts, structural markers, and anchor count. If `data/` is missing, it is fetched from `biomap-research/DeepD`.
